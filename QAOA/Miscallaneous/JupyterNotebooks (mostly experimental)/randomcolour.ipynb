```python
import os
import hashlib

value = os.uname().nodename
h = hashlib.sha256(value.encode()).hexdigest()
y = int(h, 16) % 2

print("white" if y == 1 else "wood")
print(f"Check: {y}")
```

```python
import platform
import os
import hashlib

value = platform.node() + os.uname().machine + os.uname().sysname
y = int(hashlib.sha256(value.encode()).hexdigest(), 16) % 2

print("white" if y == 1 else "wood")
print(f"Check: {y}")
```
