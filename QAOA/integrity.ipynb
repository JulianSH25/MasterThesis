```python
import os
import re
import ast

def extract_energies(text):
    pattern = r"Energies observed during COBYLA optimization:\s*(\[[^\]]*\])"
    match = re.search(pattern, text)
    if not match:
        return None
    try:
        return ast.literal_eval(match.group(1))
    except Exception:
        return None

def analyze_logs(folder, top_n=5, threshold=0.1):
    diffs = []  # (diff, filename)

    total = 0
    above_threshold = 0

    for fname in os.listdir(folder):
        if not fname.endswith(".log"):
            continue

        path = os.path.join(folder, fname)
        with open(path, "r") as f:
            content = f.read()

        energies = extract_energies(content)
        if not energies or len(energies) < 2:
            continue

        total += 1
        diff = abs(energies[0] - energies[-1])  # improvement

        if diff > threshold:
            above_threshold += 1

        diffs.append((diff, fname))

    # sort descending by improvement
    diffs.sort(reverse=True, key=lambda x: x[0])

    print(f"\nTop {top_n} improvements:")
    for d, fname in diffs[:top_n]:
        print(f"{fname}: {d:.6f}")

    print(f"\nThreshold > {threshold}: {above_threshold}/{total}")

    return diffs


# usage
analyze_logs("/Users/julian/PycharmProjects/PythonProject/MasterThesis/QAOA/logs/ADAM/20260419_154910", top_n=10, threshold=0.01)
```
